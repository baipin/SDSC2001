| Pandas Concept | Example Code | Notes |
| :--- | :--- | :--- | 
| Reading in a .csv file | ```pd.read_csv('file path')```||
| Getting column names | ```df.columns```|Assumes the data frame is named df|
|Making column names lowercase|```df.columns.str.lower()```|ie 'Name' to 'name'|
|Making column names uppercase|```df.columns.str.upper()```|ie 'Name' to 'NAME|
|Making column names capitalized|```df.columns.str.capitalize()```|ie 'name' to 'Name'|
| Getting row and column counts |```df.shape[0]``` (rows), ```df.shape[1]``` (columns)|df.shape returns a tuple of (rows, columns)|
| Getting null (missing) counts by column |```df.isnull().sum()```||
| Selecting a column by name| ```df.loc[:,'column name']``` or ```df['column name']```|An easy way to remember: think of 'loc' as short for 'location'|
|Selecting multiple columns by name|```df[['col1', 'col2',...]]```||
|Statistical summaries of a column|```df.loc[:,'column name'].mean()```|Can also use other methods ie ```.min(), .std(), .max(), etc```. Using ```.describe()``` will return multiple summary statistics at once|
| Filtering (subsetting) the data | ```df.loc[logical condition]``` | Example: ```df.loc[df['Weight'] > 10]``` will filter the data frame to only include weights greater than 10. Multiploe conditions can be specified using and/or logic within parenthesis, ie ```df.loc[(condition 1) & (condition 2)]```|
|Creating a new column|```df['New column name'] = .....```||
|Convert logical conditions to integers|```.astype(int)```|Can be used to turn true/false into 1s and 0s ie TRUE TRUE FALSE to 1 1 0|
|Grouping |```df.groupby('column name')```|Note: you can groupby multiple columns, ie ```df.groupby(['col1','col2'])```] then perform some sort of aggregation|
|Aggregation|```df.groupby().agg(['mean', 'min'])```|Will take the mean and min of all columns grouped by whatever column name is supplied to ```.groupby()```|
|Grouped summary statistics|```df.groupby('grouping column')['statistic column'].mean()```|Can use other statistical summary  methods as well. Note that using .```.loc``` after grouping for selecting a column will not work|
