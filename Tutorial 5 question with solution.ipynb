```python
# 23. Create a 7x7 matrix and reverse the rows and columns
matrix_7x7 = np.random.randint(1, 101, 49).reshape(7, 7)
print("Original 7x7 Matrix:\n", matrix_7x7)
print("----------------------------------")
# solution 1, reverse sequentially
reversed_row = matrix_7x7[::-1, :]  # reverse the rows
print("Reversed rows of 7x7 Matrix:\n", reversed_row)
print("----------------------------------")
reversed_columns = reversed_row[:, ::-1]  # reverse the columns
print("Reversed columns of 7x7 Matrix:\n", reversed_columns)
print("----------------------------------")

#solution 2, reverse simultaneously
reversed_matrix = matrix_7x7[::-1, ::-1]  # reverse the rows and columns

print("Reversed 7x7 Matrix:\n", reversed_matrix)
```
