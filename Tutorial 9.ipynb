|Reading in a .csv file (**Polars**)|```pl.read_csv('path')```|Use ```null_values = ``` to specify what the null (missing) values are encoded as in the data if necessary.|
|Reading in a .csv file (**Pandas**)|```pd.read_csv('path')```||
|Selecting a column from a data frame (**Polars**)|```df.get_column('colname')```|For inputs into matplotlib you can also use ```df.select(['colname'])```|
|Selecting a column from a data frame (**Pandas**)|```df.loc[:,'colname']```||
|Creating a histogram|```plt.hist(x, bins, alpha, color,``` <br><br> ```edgecolor)```| ```x``` = your data (ie a column of a data frame) <br><br>```bins``` = number of histogram bins <br><br> ```alpha``` = transparency of histogram bar color<br><br> ```color``` = color of the histogram bars <br><br> ```edgecolor``` = color of the lines outlining the histogram bars|
|Creating a scatterplot|```plt.plot(x,y,'o',color)```|```x``` = x-axis data (ie a column of a data frame)<br><br>```y``` = y-axis data<br><br>```'o'``` = tells Python to make a scatterplot not a lineplot<br><br>```color```= color of the points|
|Creating a lineplot|```plt.plot(x,y,label)```|```x``` = x-axis data (ie a column of a data frame)<br><br>```y``` = y-axis data<br><br>```label``` = legend label for the line <br><br> to add multiple lines use add another call to ```plt.plot()```|
|Adding labels|```plt.title('title')``` <br><br>```plt.xlabel('x-axis label')```<br><br>```plt.ylabel('y-axis label')```||
|Adding a legend|```plt.legend(loc = 'upper center')```|You don't need to specify the ```loc``` argument. Only use it if you want to control where the legend will be on the plot|
|Calculating column statistics (**Polars**)|```df.get_column('colname').max()``` ||
|Calculating column statistics (**Pandas**)|```df.loc[:,'colname'].max()```||
