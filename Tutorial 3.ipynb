- **Numbers**

    - Integers: Whole numbers, such as -3, 0, 5, 10.
    - Floating-point numbers: Numbers with decimal points, such as 3.14, -2.5, 0.0.
    - Complex numbers: Numbers with real and imaginary parts, such as 2 + 3j, -1 - 5j.
    
- **Strings:** Sequences of characters

- **Lists:** Ordered, mutable sequences of objects

- **Tuples:** Ordered, immutable sequences of objects

- **Set:** Unordered collections of unique elements

- **Booleans** True or False

- **None**
